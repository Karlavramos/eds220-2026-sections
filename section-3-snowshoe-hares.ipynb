# About the data



Metadata Exploration:
   
    1. Who assembled this data, and why?
    - Bonanza Creek Experimental Forest to analyze the fluctuation of the Snowshoe hares population because they are a 'keystone' prey species in northern boreal forests.
    
    2. What does a single row in the data represent?
    - A moment where a Hare was captured and biometric data was captured about the individual.
   
    3. When and where were the observations collected?
    - Bonanza Creek Experimental Forest, in a specific boundary box from 1999 to present.
  
    4. Does the dataset contain sensitive data?
    -no
   
    5. What study protocols were used to collect the data?
    - Capture and release
   
    6. What are the units for the hares’ weight?
    -grams
   
    7. Does the metadata say how missing data is encoded?
    -no
  
    8. Is there a publication associated with this dataset?
    - no
